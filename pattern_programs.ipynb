 ## **pattern program MATRIX**
 

```python
row=int(input("Enter the number of rows :"))
col=int(input("Enter the number of columns :"))
symbol=input("Enter the symbol :")
i=1
while i<=row:
    j=1
    while j<=col:
        print(symbol,end="")
        j=j+1
    i=i+1
    print()
```

 ## **pattern program (TRIANGLE)**

```python
row=int(input("Enter the number of rows :"))
symbol=input("Enter the symbol to use :")
i=1
while i<=row:
    j=1
    while j<=i:
        print(symbol,end="")
        j=j+1
    i=i+1
    print()
```
